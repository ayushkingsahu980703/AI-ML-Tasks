# Data Visualisation

# Interpretation: Average Bike Rentals by Season

This bar chart compares the average number of bike rentals across the four seasons. It helps identify seasonal variations in customer demand and the periods when bike-sharing services are used most frequently.

# Business insight:

- Higher average rentals indicate seasons when more bikes may be needed.

- Lower average rentals may provide opportunities for maintenance and promotional campaigns.

- Seasonal demand patterns can help businesses plan fleet availability and operating resources.

# Business recommendation:

Adjust fleet capacity and marketing strategies according to seasonal demand.

```python
#Purpose: Compare average rentals across seasons
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x='season', y='cnt', errorbar=None)


for container in ax.containers:
    ax.bar_label(container, fmt='%.1f', padding=3)

plt.title('Average Bike Rentals by Season')
plt.xlabel('Season (1: Winter, 2: Spring, 3: Summer, 4: Fall)')
plt.ylabel('Average Rentals')
plt.ylim(0, df.groupby('season')['cnt'].mean().max() * 1.15)
plt.show()
```

# 2. Average Bike Rentals by Hour

Interpretation:

This line chart shows how average bike rentals change throughout the 24 hours of a day. Peaks indicate hours with higher rental activity, while troughs indicate periods of lower demand.

# Business insight:

- Morning or evening peaks may indicate commuting-related demand.

- Lower demand during certain hours can help identify suitable maintenance periods.

- Hourly patterns help businesses anticipate when bikes are most likely to be needed.

# Business recommendation:

Position bikes in high-demand locations before peak hours and schedule maintenance during quieter periods.

```python
sns.lineplot(data=df, x='hr', y='cnt', errorbar=None, marker='o')
plt.title('Average Bike Rentals by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Average Rentals')
plt.xticks(range(0, 24))
plt.show()
```

# 3. Average Bike Rentals by Month

Interpretation:

This line chart displays average bike rentals for each month. It helps identify monthly fluctuations, seasonal trends, and periods of relatively high or low demand.

# Business insight:

- Increasing rentals may indicate periods of stronger customer activity.

- Declining rentals may indicate reduced demand during certain months.

- Monthly patterns can support planning for fleet capacity, staffing, and promotional campaigns.

# Business recommendation:

Use historical monthly demand to plan inventory and marketing activities ahead of time.

```python
sns.lineplot(data=df, x='mnth', y='cnt', errorbar=None, marker='o')
plt.title('Average Bike Rentals by Month')
plt.xlabel('Month')
plt.ylabel('Average Rentals')
plt.xticks(range(1, 13))
plt.show()
```

# 4. Registered vs. Casual Users

Interpretation:

This grouped bar chart compares the average number of registered and casual rentals on working days and non-working days. It helps reveal differences in how the two customer groups use the bike-sharing service.

# Business insight:

- Registered users may show stronger demand on working days if they use bikes for commuting.

- Casual users may show relatively stronger demand on non-working days if they use bikes for leisure.

- Comparing both customer groups helps businesses understand different usage patterns.

# Business recommendation:

Consider membership incentives for casual riders and ensure sufficient bike availability for registered users during their peak usage periods.

```python
users = df.groupby('workingday')[['registered', 'casual']].mean()
users.plot(kind='bar', figsize=(8, 5))
plt.title('Registered vs Casual Users')
plt.ylabel('Average Rentals')
plt.xlabel('Working Day (0: No, 1: Yes)')
plt.xticks(rotation=0)
plt.show()
```

# Business Insights:

# 1. Optimize bike availability during peak hours:

Hourly rental patterns can help identify periods of high demand. The business can position more bikes at busy stations before demand peaks and schedule maintenance during quieter periods.

# 3. Incorporate weather into demand forecasting:

Weather conditions may affect rental volume. Combining weather forecasts with historical demand can improve staffing and fleet allocation decisions, although the relationship should be validated.




# 4. Develop separate strategies for customer segments:

Registered users average about 154 rentals per hour, compared with about 36 for casual users. The business can focus on membership retention for registered customers and targeted membership offers for casual riders.




# 5. Improve resource allocation using demand variability:

The large variation in hourly rentals suggests that a fixed allocation of bikes may not be efficient. Demand forecasting can help reduce shortages during busy periods and underused inventory during quieter periods.
