Lists -
A list is an ordered and mutable collection in Python. It can store multiple values, including different data types

What I learned:
- Lists are ordered
- Lists are mutable
- Lists allow duplicate values
- Elements are accessed using an index
- Lists are useful when we need to maintain a collection of items

```python
#Lists

students = ["Ayush", "Rahul", "Aman"]

print(students[0])
print(students[1])
print(students[2])
```

```python
students.append("Rohit")      # Add an item
students.remove("Rahul")      # Remove an item
students[0] = "Ayush Sahu"    # Update an item
print(len(students))          # Number of items
```

Tuples:

A tuple is an ordered collection that cannot be modified after creation.


What I learned:-

- Tuples are ordered
- Tuples are immutable
- They are useful for storing data that should not be changed
- Tuple unpacking makes it easy to assign tuple values to variables

```python
#Tuples
student = ("Ayush", 23, "Data Science")

print(student[0])
print(student[2])
```

Dictionaries:

A dictionary stores data in key-value pairs.

What I learned :

- Dictionaries store key-value pairs
- Values can be accessed using keys
- Dictionaries are mutable
- They are useful for representing structured information

```python
#Dictionaries

student = {
    "id": 101,
    "name": "Ayush",
    "marks": 85
}
student
```

Sets:

A set is an unordered collection that stores only unique values.

What I learned:

- Sets store unique values
- Duplicate values are automatically removed
- Sets are useful for finding common or different elements
- Sets support operations such as union, intersection and difference.

```python
numbers = {10, 20, 30, 20, 10}

print(numbers)
```

```python
#Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
```

```python
# Student Record Management System

FILE_NAME = "students.txt"


def add_student():
    student_id = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    age = input("Enter Age: ")
    course = input("Enter Course: ")
    marks = input("Enter Marks: ")

    with open(FILE_NAME, "a") as file:
        file.write(f"{student_id}|{name}|{age}|{course}|{marks}\n")

    print("Student record added successfully!\n")


def view_students():
    try:
        with open(FILE_NAME, "r") as file:
            records = file.readlines()

        if not records:
            print("No student records found.\n")
            return

        print("\n===== Student Records =====")

        for record in records:
            student_id, name, age, course, marks = record.strip().split("|")

            print(f"ID     : {student_id}")
            print(f"Name   : {name}")
            print(f"Age    : {age}")
            print(f"Course : {course}")
            print(f"Marks  : {marks}")
            print("-" * 30)

    except FileNotFoundError:
        print("No student records found.\n")


def search_student():
    search_id = input("Enter Student ID to search: ")

    try:
        with open(FILE_NAME, "r") as file:
            records = file.readlines()

        for record in records:
            student_id, name, age, course, marks = record.strip().split("|")

            if student_id == search_id:
                print("\nStudent Found!")
                print(f"ID     : {student_id}")
                print(f"Name   : {name}")
                print(f"Age    : {age}")
                print(f"Course : {course}")
                print(f"Marks  : {marks}\n")
                return

        print("Student not found.\n")

    except FileNotFoundError:
        print("No student records found.\n")


def delete_student():
    delete_id = input("Enter Student ID to delete: ")

    try:
        with open(FILE_NAME, "r") as file:
            records = file.readlines()

        found = False

        with open(FILE_NAME, "w") as file:
            for record in records:
                student_id = record.strip().split("|")[0]

                if student_id == delete_id:
                    found = True
                else:
                    file.write(record)

        if found:
            print("Student record deleted successfully!\n")
        else:
            print("Student not found.\n")

    except FileNotFoundError:
        print("No student records found.\n")


def main():
    while True:
        print("===== Student Record Management System =====")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Delete Student")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_student()

        elif choice == "2":
            view_students()

        elif choice == "3":
            search_student()

        elif choice == "4":
            delete_student()

        elif choice == "5":
            print("Thank you for using the system!")
            break

        else:
            print("Invalid choice. Please try again.\n")


main()
```
