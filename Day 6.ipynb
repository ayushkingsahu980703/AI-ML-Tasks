**What is Data Visualisation?**

Data visualization is the process of representing data in a visual format, such as charts, graphs, and plots, to make information easier to understand and interpret.

# **Importance of Data Visualization:**




1. Makes complex data easy to understand
2. Identifies patterns and trends
3. Detects outliers and anomalies
4. Enables effective comparisons
5. Supports better decision-making

Common data visualization libraries in Python
Library:

Matplotlib :

Creating customizable charts and plots

Seaborn:

Statistical visualizations with attractive defaults.

Plotly:

Interactive charts and dashboards



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly as px
```

```python
df=pd.read_csv('/content/Cricket (1).csv',encoding='latin')
```

```python
df
```

```python
top_runs = df.nlargest(10, "Runs").sort_values("Runs", ascending=True)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_runs, x="Runs", y="Player", color="steelblue")
plt.title("Top 10 ODI Batters by Career Runs")
plt.xlabel("Career runs")
plt.ylabel("Player")
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px

top_players = df.nlargest(10, "Runs").sort_values("Runs")

fig = px.bar(
    top_players,
    x="Runs",
    y="Player",
    orientation="h",
    title="Top 10 ODI Batters by Career Runs",
    hover_data=["Mat", "Inns", "Ave", "100"]
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="Runs",
    nbins=10,
    title="Distribution of Career Runs",
    labels={"Runs": "Career Runs"},
    marginal="box"
)

fig.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Runs", bins=10, kde=True)
plt.title("Distribution of Career Runs in the Dataset")
plt.xlabel("Career runs")
plt.ylabel("Number of players")
plt.tight_layout()
plt.show()
```

```python
top_centuries = df.nlargest(5, "100").sort_values("100", ascending=False)

plt.figure(figsize=(8, 7))
plt.pie(top_centuries["100"],
        labels=top_centuries["Player"],
        autopct="%1.1f%%",
        startangle=90)
plt.title("Share of Centuries Among the Top 5 Century Scorers")
plt.tight_layout()
plt.show()
```

```python
px.line(df, x="Player", y="Runs", title="Player vs Runs", markers=True).show()
```

Sachin Tendulkar leads the dataset in career runs, with 18,426 runs.

The histogram shows that most players in this selected dataset have career totals much lower than Tendulkar's 18,426, so the distribution is pulled toward the higher end by a few high-scoring players.

Sachin Tendulkar also has the most centuries in the dataset (49), ahead of the next-highest players.

The bar chart makes the ranking of the highest run-scorers easier to compare than reading the table alone.
