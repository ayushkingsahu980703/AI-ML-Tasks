**Variables:**

Variables are names used to store values in a Python program.

**Data Types:**
Data types define the kind of value a variable can store, such as int, float, str, and bool.

**Conditional Statements: **

Conditional statements allow a program to make decisions and execute different code based on conditions.

**Loops:**

Loops repeatedly execute a block of code until a specified condition is met or for a given sequence.

**Functions:**

Functions are reusable blocks of code designed to perform a specific task.

```python
# VARIABLES

name = "Ayush"
age = 23
height = 5.8
is_student = True

print(name)
print(age)
print(height)
print(is_student)



# DATA TYPES


print(type(name))
print(type(age))
print(type(height))
print(type(is_student))


# CONDITIONAL STATEMENTS


marks = 85

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
else:
    print("C")



# LOOPS


for i in range(1, 6):
    print(i)


count = 1

while count <= 5:
    print(count)
    count += 1



# FUNCTIONS

def square(number):
    return number * number


print(square(5))


def greet(name):
    print("Hello", name)


greet("Ayush")
```

```python
#Student Grade Calculator

name= input('Enter Student Name: ')
maths=int(input('Maths Marks :'))
science=int(input('Science Marks :'))
social_science=int(input('Social Science Marks :'))
english=int(input('English Marks :'))
computer=int(input('Computer Marks :'))
total_marks=maths+science+social_science+english+computer
total_percentage = total_marks/ 5
if total_percentage >90:
  grade='A+'
elif total_percentage >80:
  grade='A'
elif total_percentage >70:
  grade='B'
elif total_percentage >60:
  grade='C'
elif total_percentage >50:
  grade='D'
else:
  grade='F'

print("\n--- Student Result ---")
print("Name:", name)
print("Total Marks:", total_marks)
print("Percentage:", total_percentage)
print("Grade:", grade)

if total_percentage >= 40:
    print("Result: PASS")
else:
    print("Result: FAIL")
```

```python
# Simple Calculator

def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b


print("Simple Calculator")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = input("Enter your choice (1-4): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == "1":
    result = add(num1, num2)

elif choice == "2":
    result = subtract(num1, num2)

elif choice == "3":
    result = multiply(num1, num2)

elif choice == "4":
    result = divide(num1, num2)

else:
    result = "Invalid choice"

print("Result:", result)
```
