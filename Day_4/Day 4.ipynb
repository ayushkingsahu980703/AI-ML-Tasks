NumPy (Numerical Python) is a Python library used for numerical computing. It provides fast, efficient multi-dimensional arrays and mathematical operations, making it one of the fundamental libraries used in Data Science, AI, and Machine Learning.

Why NumPy is important in AI/ML-

- Handles large numerical datasets efficiently
- Provides fast array operations
- Supports mathematical and statistical calculations
- Forms the foundation for libraries like Pandas, Scikit-learn and SciPy
- Makes matrix and vector operations easier
- Uses optimized operations that are generally faster than Python loops

```python
import numpy as np

# 1D Array
marks = np.array([85, 72, 90, 65, 78])

# 2D Array
data = np.array([
    [85, 72, 90],
    [65, 78, 88]
])

print(marks)
print(data)
```

```python
print(marks[0])       # First element
print(marks[-1])      # Last element
print(marks[1:4])     # Slicing
```

```python
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
```

```python
print(marks + 5)
print(marks * 2)
print(np.sqrt(marks))
```

```python
#Student Marks Analysis
marks = np.array([78, 85, 92, 67, 88, 74, 95, 81, 69, 90])

print("Marks:", marks)
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Above Average:", marks[marks > np.mean(marks)])
```

| Python List | NumPy Array |
|---|---|
| General-purpose collection | Designed for numerical computing |
| Can contain different data types | Usually contains same data type |
| Mathematical operations require loops | Supports vectorized operations |
| Generally slower for numerical operations | Faster for numerical operations |
| Less suitable for matrix operations | Excellent for matrix/vector operations |

```python
# Python List
marks = [10, 20, 30]
print([x * 2 for x in marks])

# NumPy Array
marks_np = np.array([10, 20, 30])
print(marks_np * 2)
```
