```python
df[['start','end']]=df['Span'].str.split("-",expand=True)
```

```python
df['start']=df['start'].astype(int)
df['end']=df['end'].astype(int)
```

```python
df.info()
```

```python
df['experience']=df['end']-df['start']
```

```python
df
```

```python
df=df.drop(columns=['Span','start','end'])
```

```python
df.info()
```

```python
df['HS'] = df['HS'].str.replace('*', '', regex=False)
```

```python
df['HS']=df['HS'].astype(int)
```

```python
df.info()
```

```python
df
```

```python
sns.histplot(df, x='Runs',kde=True)
plt.title('Distribution of Career Runs')
plt.ylabel('Number of Players')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(df['Ave'],kde=True, bins=12)
plt.title('Distribution of Batting Average')
plt.xlabel('Average Runs')
plt.ylabel('Number of Players')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(df['SR'],kde=True,bins=10,discrete=True)
plt.title('Distribution of Strike Rate')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['experience'],kde=True,bins=5,discrete=True)
plt.title('Distribution of Career Span')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df,
              x='100',
              color="teal",edgecolor="black",alpha=0.85)
plt.title('Distribution of Centuries',fontsize=18,
    weight="bold")
plt.xlabel('Centuries', fontsize=12,
    weight="bold")
plt.ylabel('Number of Players',fontsize=12,
    weight="bold")
plt.grid(axis="y", linestyle="--", alpha=1)
plt.show()
```
